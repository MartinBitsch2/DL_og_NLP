* try ... except er lidt ligesom if, else
* Endnu en if else: Hvis input ikke er en tekststreng eller er tom eller kun består af mellemrum, returneres 0.0

## Hvad er RAG / vektorDB / ChromaDB og hvad er fordelene

Rag står for Retrieval-Augmented Generation. Dette er bindeledet mellem LLM'en og vektorDB'en. Den henter de relevante dokumenter fra databasen og giver dem til LLM'en som kontekst inden den svarer. Den superprompter derfor LLM'en. 

I en vektordatabase struktureres ustruktureret data som text/billeder som vektorer i en database. Datapunkter som minder om hinannden vil have lignende vektorer og vil derfor ligge tæt på hinanden rent matematisk.
* Ord opbevares som betydning og ikke præcise søgeord.
